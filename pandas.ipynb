```python
import pandas as pd
import random


# OBS: Como o arquivo Excel/CSV original do curso não estava disponível para download,
# este script cria uma base sintética de 100 linhas com a mesma estrutura de colunas
# (ID, Nome, Idade, Salário, Departamento e Ativo) para possibilitar a execução
# das práticas e o acompanhamento das aulas no Pandas.

# Definição de semente para reprodutibilidade
random.seed(42)

# Listas de dados para amostragem
nomes = ['Ana', 'Carlos', 'Beatriz', 'Lucas', 'Mariana', 'Gabriel', 'Julia', 'Pedro', 'Sofia', 'Diego']
sobrenomes = ['Silva', 'Santos', 'Oliveira', 'Souza', 'Rodrigues', 'Ferreira', 'Almeida', 'Costa']
departamentos = ['Marketing', 'Recursos Humanos', 'Financeiro', 'Jurídico', 'Atendimento ao Cliente', 'Vendas']

# Construção do dicionário com 100 registros
dados = {
    'ID': [random.randint(10000, 99999) for _ in range(100)],
    'Nome': [f"{random.choice(nomes)} {random.choice(sobrenomes)}" for _ in range(100)],
    'Idade': [random.randint(21, 65) for _ in range(100)],
    'Salário': [random.randint(2000, 10000) for _ in range(100)],
    'Departamento': [random.choice(departamentos) for _ in range(100)],
    'Ativo': [random.choice([True, False]) for _ in range(100)]
}

# Criação e exportação do DataFrame
df = pd.DataFrame(dados)
df.to_csv('dados_funcionarios.csv', index=False, encoding='utf-8')
```

```python
# Aplica a negação do filtro usando o operador NOT (~)
# Inverte a lógica, retornando os funcionários com salário menor ou igual a 5000
df[~filtro_salario]
```

```python
# Para agrupar por mais de uma coluna (Série), criamos uma lista com os nomes das colunas
# e a passamos como parâmetro para o método .groupby()
grupos = ["Departamento", "Ativo"]

# Realiza o agrupamento multinível e calcula a média salarial para cada combinação
df.groupby(grupos)["Salário"].mean()
```

```python
# Utiliza o método .agg() para realizar múltiplas agregações simultâneas
# Permite renomear as colunas de saída diretamente (Named Aggregation)
df.groupby("Departamento").agg(
    idade_media=('Idade', 'mean'),
    salario_medio=('Salário', 'mean')
)
```

## DataFrame com valores nulos

```python
# OBS: Como o arquivo Excel/CSV original do curso não estava disponível para download,
# este script cria uma base sintética de 100 linhas com a mesma estrutura de colunas
# (ID, Nome, Idade, Salário, Departamento e Ativo) e valores nulos (NaN/None)
# inseridos aleatoriamente para possibilitar as práticas de tratamento de dados no Pandas.

import pandas as pd
import random
import numpy as np

# Definição de semente para reprodutibilidade
random.seed(42)
np.random.seed(42)

# Listas de dados para amostragem
nomes = ['Ana', 'Carlos', 'Beatriz', 'Lucas', 'Mariana', 'Gabriel', 'Julia', 'Pedro', 'Sofia', 'Diego']
sobrenomes = ['Silva', 'Santos', 'Oliveira', 'Souza', 'Rodrigues', 'Ferreira', 'Almeida', 'Costa']
departamentos = ['Marketing', 'Recursos Humanos', 'Financeiro', 'Jurídico', 'Atendimento ao Cliente', 'Vendas']

# Função auxiliar para introduzir valores nulos aleatoriamente (ex: ~10% de chance de ser None)
def com_nulos(valor, probabilidade_nulo=0.1):
    return None if random.random() < probabilidade_nulo else valor

# Construção do dicionário com 100 registros contendo valores nulos ocasionais
dados = {
    'ID': [random.randint(10000, 99999) for _ in range(100)],
    'Nome': [com_nulos(f"{random.choice(nomes)} {random.choice(sobrenomes)}") for _ in range(100)],
    'Idade': [com_nulos(random.randint(21, 65)) for _ in range(100)],
    'Salário': [com_nulos(random.randint(2000, 10000)) for _ in range(100)],
    'Departamento': [com_nulos(random.choice(departamentos)) for _ in range(100)],
    'Ativo': [com_nulos(random.choice([True, False])) for _ in range(100)]
}

# Criação e exportação do DataFrame
df_nulos = pd.DataFrame(dados)
df_nulos.to_csv('dados_funcionarios_com_nulos.csv', index=False, encoding='utf-8')
```

```python
df_nulos
```

```python
# Remove todas as linhas do DataFrame que possuem ao menos um valor nulo (NaN)
df_nulos.dropna()
```

```python
# Remove as linhas do DataFrame onde a coluna "Idade" contém valores nulos
df_nulos.dropna(subset=["Idade"])
```

```python
# Calcula a média da coluna "Idade" ignorando os valores nulos
media_idade = df_nulos["Idade"].mean()

# Preenche os valores nulos (NaN) da coluna "Idade" com a média calculada
df_nulos["Idade"] = df_nulos["Idade"].fillna(media_idade)
```

```python
# Preenche os valores nulos (NaN) da coluna "Idade" com a média calculada
df_nulos["Idade"].fillna(media_idade)
```

```python
#  Arredonda os valores e converte a coluna para o tipo inteiro (int)
df_nulos["Idade"] = df_nulos["Idade"].round().astype(int)
```
